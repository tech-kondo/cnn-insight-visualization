```python
from PIL import Image
import os
```

```python
def center_crop_and_resize_to_jpg(src_folder, dst_folder, size=(256, 256)):
    os.makedirs(dst_folder, exist_ok=True)  # 保存先フォルダを作成

    image_files = [f for f in os.listdir(src_folder) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
    print(f"{len(image_files)} 枚の画像を処理します...")

    for filename in image_files:
        src_path = os.path.join(src_folder, filename)
        try:
            with Image.open(src_path) as img:
                img = img.convert('RGB')  # JPG対応（RGB化）

                # 中心クロップ
                w, h = img.size
                min_side = min(w, h)
                left = (w - min_side) // 2
                top = (h - min_side) // 2
                right = left + min_side
                bottom = top + min_side
                img_cropped = img.crop((left, top, right, bottom))

                # リサイズ
                img_resized = img_cropped.resize(size)

                # 保存先ファイル名を .jpg に統一
                base_name = os.path.splitext(filename)[0]
                new_path = os.path.join(dst_folder, base_name + ".jpg")
                img_resized.save(new_path, format='JPEG')

        except Exception as e:
            print(f"{filename} の処理中にエラー発生：{e}")

    print("✅ 中心クロップ＋リサイズ＋JPG保存（別フォルダ）完了！")
```

```python
src_folder='/content/drive/MyDrive/cnn_visualization_workspace/data/biglove_fam_data'
dst_folder=src_folder + '_crop'
```

```python
center_crop_and_resize_to_jpg(src_folder=src_folder, dst_folder=dst_folder)
```
