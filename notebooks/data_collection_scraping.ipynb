```python
# ライブラリのインストール
!pip install icrawler
# ライブラリのインポート
from icrawler.builtin import BingImageCrawler, GoogleImageCrawler
import os
```

```python
def download_dog_images(breed_name, num_images=50):
    try:
      crawler = BingImageCrawler(storage={'root_dir': f'data/collection/Bing/{breed_name}'})
      crawler.crawl(keyword=breed_name + " dog", max_num=num_images)
    except Exception as e:
      print(f"エラー発生：{e}")
```

```python
os.makedirs('data/collection/Bing/Dachshund', exist_ok=True)
for breed in ["Dachshund"]:
    download_dog_images(breed, num_images=300)
```
